```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib as mpl
import imageio_ffmpeg

from PIL import Image
from pathlib import Path
from matplotlib.animation import FuncAnimation, FFMpegWriter
from matplotlib.offsetbox import OffsetImage, AnnotationBbox

# FILE
CSV_FILE = "browser-ww-monthly-200901-202609-cleaned-main-browsers.csv"

# SETTINGS — short test first
FPS = 30
DURATION = 330
FRAMES = FPS * DURATION

# FULL HD
WIDTH = 1920
HEIGHT = 1080
DPI = 100

# DATA
df = pd.read_csv(CSV_FILE)
df["Date"] = pd.to_datetime(df["Date"])
df = df.sort_values("Date").reset_index(drop=True)

browsers = [
    "Chrome", "IE", "Firefox", "Safari",
    "Opera", "Edge", "Edge Legacy",
    "Brave", "Yandex Browser", "Other"
]

colors = {
    "Chrome": "#34A853",
    "IE": "#1784D5",
    "Firefox": "#FF7139",
    "Safari": "#76B9ED",
    "Opera": "#E53935",
    "Edge": "#15B8A6",
    "Edge Legacy": "#3367C1",
    "Brave": "#FB542B",
    "Yandex Browser": "#E34F59",
    "Other": "#81899D"
}

display_names = {
    "IE": "Internet Explorer"
}

# LOGOS
logo_files = {
    "Chrome": "chrome.png",
    "IE": "ie.png",
    "Firefox": "firefox.png",
    "Safari": "safari.png",
    "Opera": "opera.png",
    "Edge": "edge.png",
    "Edge Legacy": "edge_legacy.png",
    "Brave": "brave.png",
    "Yandex Browser": "yandex.png"
}

browser_logos = {}

for browser, filename in logo_files.items():
    image = Image.open(filename).convert("RGBA")
    image.thumbnail((64, 64))

    canvas = Image.new(
        "RGBA", (64, 64), (0, 0, 0, 0)
    )

    x = (64 - image.width) // 2
    y = (64 - image.height) // 2

    canvas.paste(image, (x, y), image)
    browser_logos[browser] = np.array(canvas)

values = df[browsers].to_numpy(dtype=float)
dates = df["Date"]

print("Dataset months:", len(df))
print("Logos loaded:", len(browser_logos))
print("Animation duration:", DURATION, "seconds")
print("Frames:", FRAMES)
```

```python

from matplotlib.animation import FuncAnimation
from matplotlib.offsetbox import OffsetImage, AnnotationBbox

fig = plt.figure(figsize=(19.2, 10.8), dpi=100)
fig.patch.set_facecolor("#0D1625")

# Reserve sufficient room for all labels
ax = fig.add_axes([0.28, 0.16, 0.65, 0.66])
ax.set_facecolor("#0D1625")

# Figure-level elements created once
fig.text(
    0.05, 0.95, "THE BROWSER WARS",
    fontsize=32, fontweight="bold",
    color="white", va="top"
)

fig.text(
    0.05, 0.89,
    "Worldwide Desktop Browser Usage Share | 2009–2026",
    fontsize=16, color="#A9B7CB"
)

date_text = fig.text(
    0.93, 0.08, "",
    fontsize=28, fontweight="bold",
    color="#F2BE68", ha="right"
)

fig.text(
    0.05, 0.05,
    "SOURCE: STATCOUNTER GLOBAL STATS",
    fontsize=12, color="#A9B7CB"
)

milestone_text = fig.text(
    0.55, 0.86, "",
    fontsize=14, fontweight="bold",
    ha="center", color="#F2BE68"
)

# For smoother transitions, calculate fractional rank
def fractional_ranks(shares):
    order = np.argsort(-shares)
    ranks = np.empty(len(shares))
    ranks[order] = np.arange(len(shares))
    return ranks

rank_positions = np.array([
    fractional_ranks(row) for row in values
])

def update(frame):
    ax.clear()
    ax.set_facecolor("#0D1625")

    # Interpolate through the 213 months
    position = frame / (FRAMES - 1) * (len(df) - 1)
    index = min(int(position), len(df) - 2)
    fraction = position - index

    current = (
        values[index] * (1 - fraction)
        + values[index + 1] * fraction
    )

    # Interpolate positions for smoother swaps
    current_ranks = (
        rank_positions[index] * (1 - fraction)
        + rank_positions[index + 1] * fraction
    )

    for i, browser in enumerate(browsers):
        share = current[i]
        y = current_ranks[i]

        ax.barh(
            y, share,
            height=0.65,
            color=colors[browser],
            zorder=2
        )

        # Browser name: fixed column
        ax.text(
            -11, y,
            display_names.get(browser, browser),
            color="white",
            fontsize=15,
            fontweight="bold",
            va="center",
            ha="right"
        )

        # Logo beside the name/bar
        if browser in browser_logos:
            imagebox = OffsetImage(
                browser_logos[browser],
                zoom=0.58
            )

            ab = AnnotationBbox(
                imagebox,
                (-5, y),
                frameon=False
            )

            ax.add_artist(ab)

        # Market share percentage
        ax.text(
            min(share + 0.7, 85),
            y,
            f"{share:.2f}%",
            color="white",
            fontsize=13,
            va="center",
            fontweight="bold"
        )

    ax.set_ylim(9.7, -0.7)
    ax.set_xlim(-19, 90)

    ax.set_xticks([0, 20, 40, 60, 80])
    ax.set_xticklabels(
        ["0%", "20%", "40%", "60%", "80%"],
        color="#A9B7CB",
        fontsize=12
    )

    ax.set_yticks([])
    ax.grid(axis="x", alpha=0.12)
    ax.set_axisbelow(True)

    for spine in ax.spines.values():
        spine.set_visible(False)

    current_date = dates.iloc[index]
    date_text.set_text(current_date.strftime("%b %Y"))

    if pd.Timestamp("2012-05-01") <= current_date <= pd.Timestamp("2012-06-01"):
        milestone_text.set_text(
            "CHROME OVERTAKES INTERNET EXPLORER"
        )
    else:
        milestone_text.set_text("")

animation_preview = FuncAnimation(
    fig,
    update,
    frames=FRAMES,
    interval=1000 / FPS,
    repeat=False,
    cache_frame_data=False
)

plt.close(fig)

print("Animation created successfully!")
```

```python

import matplotlib as mpl
import imageio_ffmpeg
from matplotlib.animation import FFMpegWriter

mpl.rcParams["animation.ffmpeg_path"] = (
    imageio_ffmpeg.get_ffmpeg_exe()
)

writer = FFMpegWriter(
    fps=FPS,
    bitrate=8000
)

animation_preview.save(
    "Browser_Wars_Final_5m30s_1080p.mp4",
    writer=writer,
    dpi=100
)
print("Video exported successfully!")
```
